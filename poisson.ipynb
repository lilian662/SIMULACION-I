#Introducción
La distribución de Poisson es una distribución de probabilidad discreta que se utiliza para modelar el número de veces que ocurre un evento dentro de un intervalo fijo de tiempo, espacio o cualquier unidad de observación. Esta distribución supone que los eventos ocurren de manera independiente y con una tasa promedio constante.

Una variable aleatoria $X$ sigue una distribución Poisson con parámetro $\lambda$ si su función de probabilidad está dada por:

$$
P(X=i)=\frac{e^{-\lambda}\lambda^i}{i!}, \qquad i=0,1,2,\dots
$$

donde $\lambda$ representa el número promedio esperado de ocurrencias del evento.

La distribución de Poisson tiene múltiples aplicaciones en distintas áreas, ya que permite modelar fenómenos como el número de llamadas recibidas en una central telefónica, la cantidad de clientes que llegan a un establecimiento, el número de accidentes en una carretera, defectos en procesos de manufactura y fallas en sistemas, entre otros.



```python
import numpy as np

def poisson_inversa(lam):
    U = np.random.uniform(0,1)

    i = 0
    p = np.exp(-lam)
    F = p

    while U > F:
        p = (lam*p)/(i+1)
        F = F + p
        i = i + 1

    return i


lam = 3

n = 10000
muestra = [poisson_inversa(lam) for _ in range(n)]

print(muestra)
```
